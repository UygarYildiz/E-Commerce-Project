```python
import pandas as pd
import os
```

```python
raw_data_path=os.path.join("..","data","raw")

files=[f for f in os.listdir(raw_data_path)]
print(files)
```

```python
data_frames={}

for file in files:
    df_name=file.replace(".csv","")
    file_path=os.path.join(raw_data_path,file)
    data_frames[df_name]=pd.read_csv(file_path)
```

```python
for name,df in data_frames.items():
    print(f"---{name}---")
    print(df.dtypes)
    print("\n")
```

```python
# Veri setinin incelenmesis ve E-R diyagramının çıkarılabilmesi için genel bir veri analizi.
def analyzte_dataframe(name,df):
    print(f"-"*50)
    print(f"Tablo: {name.upper()}")
    print(f"Satır Sayısı: {df.shape[0]}")
    print(f"-"*50)
    print(f"{"Sütun Adı":<35} | {"Tip":<10} | {"Dolu %":<8} | {"Unique":<8} | {"Örnek Veri"}")
    for col in df.columns:
        # Veri tiplerini anlamak için boş olmayan bir örnek veri alıyoruz.
        sample=df[col].dropna().iloc[0] if not df[col].dropna().empty else "Dataframe Boş"

        # Kategorik değişken belirlemek için benzersiz değerlerin sayısı
        unique_count=df[col].nunique()
        dtype=df[col].dtype

        # Sütünun doluluk oranı ?. Dropna mi yoksa fill mi yapılacağını belirlemek için.
        non_null_count=df[col].shape[0]
        doluluk=(non_null_count / len(df))*100

        # Sütünun primary key olup olmadığının belirlenmesi
        is_pk="Primary Key" if unique_count==df.shape[0] else ""

        print(f"{col:<35} | {str(dtype):10} | {doluluk:>5.1f}% |{unique_count:<8} | {str(sample):<20} {is_pk}")



   
        

          
        
        
```

```python
for name,df in data_frames.items():
    analyzte_dataframe(name,df)
    print("\n")
    
    
```
